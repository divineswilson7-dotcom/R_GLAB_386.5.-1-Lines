```python
# Real Estate Media Example:
# Does Better Marketing Help Homes Sell Faster?

listing_number = [1, 2, 3, 4, 5, 6, 7]

basic_photos = [28, 25, 24, 21, 23, 19, 20]
premium_media = [20, 18, 15, 14, 12, 10, 8]

# Basic photography package
plt.plot(
    listing_number,
    basic_photos,
    linestyle='--',
    marker='s',
    label='Basic Photos'
)

# Premium media package
plt.plot(
    listing_number,
    premium_media,
    linestyle=':',
    marker='D',
    label='Photos + Floor Plan + 3D Tour'
)

plt.title('Real Estate Marketing vs. Days on Market')
plt.xlabel('Listing Number')
plt.ylabel('Days on Market')

plt.legend()
plt.grid(alpha=0.25)

plt.show()
```

```python
# Creative Real Estate Plot: House Markers

years = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

average_price = [245000, 265000, 295000, 310000, 325000, 340000, 355000]

# Plot the line
plt.plot(
    years,
    average_price,
    linestyle='--',
    linewidth=2
)

# Put a house at each data point
for year, price in zip(years, average_price):
    plt.text(
        year,
        price,
        '🏠',
        fontsize=18,
        ha='center',
        va='center'
    )

plt.title('Home Values Over Time')
plt.xlabel('Year')
plt.ylabel('Average Home Price ($)')

plt.grid(linestyle=':', alpha=0.3)

plt.show()
```
